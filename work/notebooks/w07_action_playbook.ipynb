## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

```python
import pandas as pd
import numpy as np

# Generate mock data
data = {
    'url': ['/blog/seo-guide', '/blog/web-dev-trends', '/tutorial/react-basics', '/news/tech-update', '/guide/iot-setup'],
    'refresh_score': [0.92, 0.75, 0.65, 0.45, 0.12],
    'current_traffic': [1200, 850, 1500, 3000, 500]
}
df_results = pd.DataFrame(data)

# Create Reason Codes
conditions = [
    (df_results['refresh_score'] > 0.8),
    (df_results['refresh_score'] > 0.5) & (df_results['refresh_score'] <= 0.8),
    (df_results['refresh_score'] <= 0.5)
]
choices = [
    'URGENT REFRESH: Traffic decaying rapidly',
    'MONITOR/MINOR UPDATE: Stagnant momentum',
    'NO ACTION: Content performing well'
]
df_results['reason_code'] = np.select(conditions, choices, default='UNKNOWN')

# Filter action queue
action_queue = df_results[df_results['refresh_score'] > 0.5].sort_values(by='refresh_score', ascending=False)

# Pakai print(..., index=False) biar bentuknya rapi dan tidak nempel-nempel
print(action_queue[['url', 'refresh_score', 'reason_code']].to_string(index=False))
```

This table translates the model's output probabilities into actionable directives. We prioritize pages with the highest refresh scores, assigning clear reason codes so the content team understands exactly why a page was flagged for an update.

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

- **Intended Use:** This tool is designed strictly for decision-support to help content editors and SEO managers prioritize which existing pages to update based on traffic decay patterns.
- **Limits:** The model relies purely on historical quantitative metrics. It cannot evaluate factual accuracy or predict sudden shifts in external user search intent.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

- **Human Review Required:** An editor must manually verify flagged URLs to ensure the content is genuinely outdated rather than affected by seasonal trends.
- **The No-Go List:** Direct automation connecting this queue to a generative AI pipeline for unmonitored publishing is strictly forbidden.

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

The recommendations will go stale and require model retraining if:
1. A major Google Core Algorithm Update disrupts baseline SERP structures.
2. The target website undergoes a massive URL migration or structural overhaul.
